### For balance the data we will implement Undersampling method

#### Undersampling

```python
# Separate the majority and minority classes
data_majority = data[data.isFraud == 0]
data_minority = data[data.isFraud == 1]
```

```python
# Undersample the majority class
data_majority_undersampled = resample(data_majority,
                                      replace=False,     # sample without replacement
                                      n_samples=len(data_minority),    # to match minority class
                                      random_state=42)   # reproducible results

# Combine the undersampled majority class with the minority class
data_undersampled = pd.concat([data_majority_undersampled, data_minority])
```

#### DNN - deep neural network

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout

# Define the model
dnn = Sequential([
    Dense(256, activation='relu', input_shape=(X_train_under.shape[1],)),  # Adjust the input_shape based on your feature size
    Dropout(0.3),  # Dropout layer to reduce overfitting
    Dense(128, activation='relu'),
    Dropout(0.3),
    Dense(64, activation='relu'),
    Dropout(0.2),
    Dense(1, activation='sigmoid')  # Output layer with sigmoid activation for binary classification
])

# Compile the model
dnn.compile(
    optimizer='adam',  # Optimizer
    loss='binary_crossentropy',  # Loss function for binary classification
    metrics=['accuracy']  # Metric to monitor
)

# Model summary
dnn.summary()

# Train the model
history = dnn.fit(
    X_train_under, y_train_under,
    validation_data=(X_test_under, y_test_under),
    epochs=10,  # Number of epochs
    batch_size=32,  # Batch size
    verbose=1
)

# Predictions (probabilities -- output of the sigmoid function)
y_pred_prob = dnn.predict(X_test_under)
y_pred = (y_pred_prob > 0.5).astype(int)  # Convert probabilities to binary predictions
```

```python
# Evaluate the model
loss, accuracy = dnn.evaluate(X_test_under, y_test_under, verbose=1)
print(f"Test Accuracy: {accuracy * 100:.2f}%")
```

```python
# Classification report
dnn_cm = classification_report(y_test_under, y_pred)
print(dnn_cm)
```

```python
# Confusion matrix
cm = confusion_matrix(y_test_under, y_pred)
print("DNN confusion matrix :\n",cm)

# Plotting confusion matrix
plt.figure(figsize=(10, 5))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=['Negative', 'Positive'], yticklabels=['Negative', 'Positive'])
plt.xlabel('Predicted Label')
plt.ylabel('True Label')
plt.title('Confusion Matrix')
plt.show()
```

```python
import shap
import pandas as pd
import numpy as np
import tensorflow as tf

# Take 10% sample for SHAP calculation
X_train_under_sample = X_train_under.sample(frac=0.10, random_state=42)
X_test_under_sample = X_test_under.sample(frac=0.10, random_state=42)

# Ensure data is float32 for Keras
X_train_under_sample_np = X_train_under_sample.values.astype(np.float32)
X_test_under_sample_np = X_test_under_sample.values.astype(np.float32)

# Initialize DeepExplainer
explainer = shap.DeepExplainer(dnn, X_train_under_sample_np)

# Calculate SHAP values - returns a list for multi-output models
shap_values = explainer.shap_values(X_test_under_sample_np)

# For binary classification with DeepExplainer, shap_values is often a list [val_class0, val_class1]
# We use the values for the positive class (fraud)
if isinstance(shap_values, list):
    shap_to_plot = shap_values[1]
else:
    shap_to_plot = shap_values

# Visualize SHAP values for the positive class
shap.summary_plot(shap_to_plot, X_test_under_sample)
```

### Recurrent Neural Network RNN

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import SimpleRNN, Dense

# Define the RNN model
rnn = Sequential()
rnn.add(SimpleRNN(50, input_shape=(X_train_under.shape[1], 1), return_sequences=True))
rnn.add(SimpleRNN(50))
rnn.add(Dense(1, activation='sigmoid'))

# Compile the model
rnn.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])


# Convert DataFrame to numpy array before reshaping
X_train_under_reshaped = np.array(X_train_under).reshape(X_train_under.shape[0], X_train_under.shape[1], 1)
X_test_under_reshaped = np.array(X_test_under).reshape(X_test_under.shape[0], X_test_under.shape[1], 1)


his = rnn.fit(X_train_under_reshaped, y_train_under, epochs=20, batch_size=64, validation_split=0.2)

# Predict the classes
y_pred = rnn.predict(X_test_under_reshaped)
y_pred_classes = (y_pred > 0.5).astype(int)
```

```python
# Calculate classification report and confusion matrix
clf_report = classification_report(y_test_under, y_pred_classes)
conf_matrix = confusion_matrix(y_test_under, y_pred_classes)
accuracy = accuracy_score(y_test_under, y_pred_classes)


print("Accuracy:", accuracy)
print('='*60)
print("Confusion Matrix:\n", conf_matrix)
print('='*60)
print("Classification Report:\n", clf_report)
```

```python
# Plotting training and validation accuracy
plt.plot(his.history['accuracy'], label='train accuracy')
plt.plot(his.history['val_accuracy'], label='validation accuracy')
plt.title('Model accuracy')
plt.ylabel('Accuracy')
plt.xlabel('Epoch')
plt.legend(loc='upper left')
plt.show()
```

### Roc_curve, auc for Existing System

```python

from sklearn.metrics import roc_curve, auc

y_pred_prob_ann = dnn.predict(X_test_under)  # Predicted probabilities for ANN (DNN)

# RNN model
y_pred_prob_rnn = rnn.predict(X_test_under_reshaped)  # Predicted probabilities for RNN


y_pred_prob_sgd = sgd.decision_function(X_test_under)  # Get decision function scores

# Step 2: Calculate ROC and AUC for each model
# ROC and AUC for ANN
fpr_ann, tpr_ann, _ = roc_curve(y_test_under, y_pred_prob_ann)
roc_auc_ann = auc(fpr_ann, tpr_ann)

# ROC and AUC for RNN
fpr_rnn, tpr_rnn, _ = roc_curve(y_test_under, y_pred_prob_rnn)
roc_auc_rnn = auc(fpr_rnn, tpr_rnn)

# ROC and AUC for SGD
fpr_sgd, tpr_sgd, _ = roc_curve(y_test_under, y_pred_prob_sgd)
roc_auc_sgd = auc(fpr_sgd, tpr_sgd)

# Step 3: Plot the ROC curve for all models
plt.figure(figsize=(8, 6))

# Plot for ANN
plt.plot(fpr_ann, tpr_ann, label=f'ANN (AUC = {roc_auc_ann:.2f})')

# Plot for RNN
plt.plot(fpr_rnn, tpr_rnn, label=f'RNN (AUC = {roc_auc_rnn:.2f})')

# Plot for SGD
plt.plot(fpr_sgd, tpr_sgd, label=f'SGD (AUC = {roc_auc_sgd:.2f})')

# Plot the diagonal line for random guess
plt.plot([0, 1], [0, 1], 'k--', label='Random Guess')

# Plot settings
plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('ROC Curve for ANN, RNN, and SGD')
plt.legend(loc='lower right')
plt.grid(True)
plt.show()
```

```python
import numpy as np
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report

def federated_averaging(X_train, y_train, X_test, y_test, n_clients=5):
    # Split data among clients
    X_clients = np.array_split(X_train, n_clients)
    y_clients = np.array_split(y_train, n_clients)

    client_models = []

    # Local training phase
    print(f'Training on {n_clients} clients...')
    for i in range(n_clients):
        model = RandomForestClassifier(n_estimators=100, random_state=42)
        model.fit(X_clients[i], y_clients[i])
        client_models.append(model)
        print(f'Client {i+1} training complete.')

    # Aggregation Phase (Simulation via weight/prediction averaging)
    # In a real FL scenario, we'd average weights. Here we average prediction probabilities.
    all_probs = []
    for model in client_models:
        probs = model.predict_proba(X_test)
        all_probs.append(probs)

    avg_probs = np.mean(all_probs, axis=0)
    final_preds = np.argmax(avg_probs, axis=1)

    # Evaluate
    acc = accuracy_score(y_test, final_preds)
    return acc, final_preds

# Execute Federated Learning Simulation
fl_accuracy, fl_preds = federated_averaging(X_train_under, y_train_under, X_test_under, y_test_under)

print(f'\nFederated Learning Model Accuracy: {fl_accuracy*100:.2f}%')
print('\nClassification Report:')
print(classification_report(y_test_under, fl_preds))
```

```python
# Define the prediction function using the aggregated FL model logic
def predict_fraud(input_data):
    # Extract values in correct order for model: type, amount, oldbalanceOrg, newbalanceOrig, oldbalanceDest, newbalanceDest
    # Note: 'type' must be label encoded as previously defined (TRANSFER=4, etc.)
    type_mapping = {'CASH_IN': 0, 'CASH_OUT': 1, 'DEBIT': 2, 'PAYMENT': 3, 'TRANSFER': 4}

    encoded_type = type_mapping.get(input_data['type'], 3) # Default to payment if unknown
    features = np.array([[encoded_type, input_data['amount'], input_data['oldbalanceOrg'],
                          input_data['newbalanceOrig'], input_data['oldbalanceDest'], input_data['newbalanceDest']]])

    # Use the averaged probabilities from our client models (Ensemble approach)
    all_probs = []
    for model in client_models:
        probs = model.predict_proba(features)
        all_probs.append(probs)

    avg_probs = np.mean(all_probs, axis=0)
    prediction = np.argmax(avg_probs, axis=1)[0]

    return "Fraud Transaction" if prediction == 1 else "No Fraud Transaction"

# Example input values
input_values = {
    "step": 1,
    "type": "TRANSFER",
    "amount": 1000.0,
    "oldbalanceOrg": 1500.0,
    "newbalanceOrig": 500.0,
    "oldbalanceDest": 0.0,
    "newbalanceDest": 1000.0
}

# Predict dynamically
fraud_prediction = predict_fraud(input_values)
print(f"Prediction for input_values: {fraud_prediction}")
```

## Complete Code: Centralized vs Federated Comparison

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier,VotingClassifier
from xgboost import XGBClassifier
from catboost import CatBoostClassifier

from sklearn.metrics import accuracy_score

# ---------------------------------------------------
# Models Dictionary
# ---------------------------------------------------

models = {
    "Decision Tree": DecisionTreeClassifier(random_state=42),

    "Random Forest": RandomForestClassifier(
        n_estimators=200,
        random_state=42
    ),

    "Gradient Boosting": GradientBoostingClassifier(
        n_estimators=200
    ),

    "XGBoost": XGBClassifier(
        n_estimators=300,
        learning_rate=0.05,
        max_depth=6,
        eval_metric='logloss',
        random_state=42
    ),

    "CatBoost": CatBoostClassifier(
        iterations=300,
        learning_rate=0.05,
        depth=6,
        verbose=0,
        random_state=42
    )
}

# ---------------------------------------------------
# CENTRALIZED TRAINING
# ---------------------------------------------------

centralized_results = {}

for name, model in models.items():

    model.fit(X_train_under, y_train_under)

    y_pred = model.predict(X_test_under)

    acc = accuracy_score(y_test_under, y_pred)

    centralized_results[name] = acc

# ---------------------------------------------------
# SIMULATED FEDERATED LEARNING
# ---------------------------------------------------

num_clients = 3

client_X = np.array_split(X_train_under, num_clients)
client_y = np.array_split(y_train_under, num_clients)

federated_results = {}

for name, model in models.items():

    client_predictions = []

    for i in range(num_clients):

        local_model = model.__class__(**model.get_params())

        local_model.fit(client_X[i], client_y[i])

        preds = local_model.predict(X_test_under)

        client_predictions.append(preds)

    client_predictions = np.array(client_predictions)

    federated_pred = np.round(np.mean(client_predictions, axis=0))

    acc = accuracy_score(y_test_under, federated_pred)

    federated_results[name] = acc

# ---------------------------------------------------
# RESULTS TABLE
# ---------------------------------------------------

results_df = pd.DataFrame({
    "Model": list(models.keys()),
    "Centralized Accuracy": list(centralized_results.values()),
    "Federated Accuracy": list(federated_results.values())
})

print(results_df)

# ---------------------------------------------------
# COMPARISON PLOT
# ---------------------------------------------------

x = np.arange(len(results_df["Model"]))

width = 0.35

plt.figure(figsize=(10,6))

plt.bar(x - width/2, results_df["Centralized Accuracy"], width, label="Centralized")

plt.bar(x + width/2, results_df["Federated Accuracy"], width, label="Federated")

plt.xticks(x, results_df["Model"], rotation=20)

plt.ylabel("Accuracy")

plt.title("Centralized vs Federated Learning Model Performance")

plt.legend()

plt.tight_layout()

plt.show()
```
